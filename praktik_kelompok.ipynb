**Analisis Awal**

Kami menggunakan dataset nilai akademik siswa. Dataset ini berisi data siswa, kelas, mata pelajaran, nilai, jenis ujian, tanggal ujian, dan guru pengampu.

Dari dataset ini kami ingin mengetahui bagaimana kondisi nilai siswa dan perbedaan nilai berdasarkan mata pelajaran dan kelas.

**Pertanyaan Analisis**

1. Berapa rata-rata nilai siswa pada setiap mata pelajaran?
2. Mata pelajaran apa yang memiliki rata-rata nilai paling tinggi dan paling rendah?
3. Bagaimana perbandingan rata-rata nilai dari setiap kelas?

**Dugaan Masalah Kualitas Data**

Sebelum data dianalisis, kami akan mengecek terlebih dahulu apakah ada masalah pada dataset.

Beberapa masalah yang ditemukan atau kemungkinan ditemukan yaitu:

* Ada data yang kosong atau missing value.
* Ada data yang duplikat.
* Ada tipe data yang belum sesuai.
* Ada format tanggal yang berbeda.
* Ada penulisan nilai yang tidak sama, seperti angka yang memiliki tulisan "poin".
* Ada nilai yang tidak sesuai dengan nilai akademik.

```python
import pandas as pd
 # Data Loading
df = pd.read_csv('dataset_nilai_akademik.csv')

print("HEAD")
print(df.head())

print(" INFO ")
print(df.info())

print(" DESCRIBE ")
print(df.describe())

print(" SHAPE ")
print(df.shape)
```

```python
 # Inspection

print(" MISSING VALUE ")
print(df.isnull().sum())

print(" DUPLIKAT ")
print(df.duplicated().sum())

print(" TIPE DATA ")
print(df.dtypes)
```

```python
# DATA CLEANING

# Membersihkan tulisan "poin"
df['nilai'] = df['nilai'].astype(str).str.replace(
    ' poin', '', regex=False
)

# Mengubah nilai menjadi angka
df['nilai'] = pd.to_numeric(df['nilai'], errors='coerce')

# Mengubah nilai tidak valid menjadi missing value
df.loc[df['nilai'] > 100, 'nilai'] = pd.NA

# Mengisi missing value nilai dengan median
median_nilai = df['nilai'].median()
df['nilai'] = df['nilai'].fillna(median_nilai)

# Mengisi missing value guru
df['guru_pengampu'] = df['guru_pengampu'].fillna(
    'Tidak diketahui'
)

# Menghapus duplikat
df = df.drop_duplicates()

# Mengubah tanggal menjadi datetime
df['tanggal_ujian'] = pd.to_datetime(
    df['tanggal_ujian'],
    dayfirst=True,
    errors='coerce'
)

# Menyamakan penulisan jenis ujian
df['jenis_ujian'] = df['jenis_ujian'].str.upper()
```

```python
# CEK SETELAH CLEANING
# =========================

print("=== SETELAH CLEANING ===")

print(" Missing value:")
print(df.isnull().sum())

print(" Duplikat:")
print(df.duplicated().sum())

print(" Tipe data:")
print(df.dtypes)

print(" Jumlah data:")
print(df.shape)
```

```python
# DATA MANIPULATION
# =========================

# FILTERING
print(" NILAI >= 75 ")

nilai_tinggi = df[df['nilai'] >= 75]

print(
    nilai_tinggi[
        ['nama', 'kelas', 'mata_pelajaran', 'nilai']
    ]
)


# SORTING
print(" NILAI TERTINGGI ")

nilai_urut = df.sort_values(
    by='nilai',
    ascending=False
)

print(
    nilai_urut[
        ['nama', 'kelas', 'mata_pelajaran', 'nilai']
    ].head(10)
)


# KOLOM TURUNAN
def kategori_nilai(nilai):
    if nilai >= 90:
        return 'Sangat Baik'
    elif nilai >= 80:
        return 'Baik'
    elif nilai >= 75:
        return 'Cukup'
    else:
        return 'Perlu Perbaikan'


df['kategori_nilai'] = df['nilai'].apply(
    kategori_nilai
)

print(" KATEGORI NILAI ")

print(
    df[
        ['nama', 'nilai', 'kategori_nilai']
    ].head(10)
)


# GROUPBY MATA PELAJARAN
print(" RATA-RATA NILAI PER MATA PELAJARAN ")

rata_mapel = df.groupby(
    'mata_pelajaran'
)['nilai'].mean()

print(rata_mapel)


# GROUPBY KELAS
print(" RATA-RATA NILAI PER KELAS ")

rata_kelas = df.groupby(
    'kelas'
)['nilai'].mean()

print(rata_kelas)


# GROUPBY KELAS DAN MATA PELAJARAN
print(" RATA-RATA NILAI PER KELAS DAN MAPEL ")

rata_mapel_kelas = df.groupby(
    ['kelas', 'mata_pelajaran']
)['nilai'].mean()

print(rata_mapel_kelas)

```

```python
# SIMPAN DATASET BERSIH

df.to_csv(
    'dataset_bersih.csv',
    index=False
)

print("Dataset bersih berhasil disimpan.")
```
